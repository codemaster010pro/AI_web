```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()

llm = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.3)
```

```python
from pydantic import BaseModel
class choice(BaseModel):
    name:str
    choices:str
```

```python
def game(state:choice):
    current_name = state.name
    current_choices = state.choices
    response = llm.invoke(f"my name is {current_name}.my message is choose one from options given{current_choices}").content
    state.choices = f"your choices was {current_choices}. my response is {response} "
    return state
```

```python
from langgraph.graph import StateGraph,END,START

graph = StateGraph(choice)

graph.add_node("game",game)
graph.add_edge(START,"game")
graph.add_edge("game",END)

graph_1 = graph.compile()
graph_1.invoke({"name":"dhruvansh","choices":"langgraph or langchain"})
```
